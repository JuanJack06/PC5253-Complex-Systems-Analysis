```python
#PC5253 Complex System Analysis
#Assignment 1

#Question 3 (no AI assistance was used)
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
#Generate n random x_i from Poisson distribution
rand = np.random.default_rng()
def poisson(l,n):
    return rand.poisson(l,n)

#Calculate the scaled sample average for X, the list of all x_i generated
def sample_avg(n,X):
    return np.sqrt(n)*sum(X)

l = 2
```

```python
# n = 1
n = 1
X_avg_1 = np.linspace(0,1,100)
#Repeat the generation of new x_i and calculation of scaled sample average to obtain a distribution
for i in range(100):
    X = poisson(l,n)
    X_avg_1[i] = sample_avg(n, X)

# n = 10
n = 10
X_avg_10 = np.linspace(0,1,100)
#Repeat the generation of new x_i and calculation of scaled sample average to obtain a distribution
for i in range(100):
    X = poisson(l,n)
    X_avg_10[i] = sample_avg(n, X)

# n = 100
n = 100
X_avg_100 = np.linspace(0,1,100)
#Repeat the generation of new x_i and calculation of scaled sample average to obtain a distribution
for i in range(100):
    X = poisson(l,n)
    X_avg_100[i] = sample_avg(n, X)
```

The mean value and variance used in the corresponding normal distribution is the mean and variance of the generated list of scaled sample variance.

```python
#plot n=1
mean = np.mean(X_avg_1)
sd = np.std(X_avg_1)
t_1 = np.linspace(mean-3*sd,mean+3*sd,1000)
distribution_1 = norm.pdf(t_1,mean,sd)

plt.hist(X_avg_1, color = 'red', density = True)
plt.plot(t_1,distribution_1, color = 'black')
plt.xlabel('Sample Average')
plt.ylabel('Probability (density)')
plt.title('Histogram of sample average at n=1 and corresponding normal distribution')
```

```python
#plot n=10
mean = np.mean(X_avg_10)
sd = np.std(X_avg_10)
t_10 = np.linspace(mean-3*sd,mean+3*sd,1000)
distribution_10 = norm.pdf(t_10,mean,sd)

plt.hist(X_avg_10, color = 'blue', density = True)
plt.plot(t_10,distribution_10, color = 'black')
plt.xlabel('Sample Average')
plt.ylabel('Probability (density)')
plt.title('Histogram of sample average at n=10 and corresponding normal distribution')
```

```python
#plot n=100
mean = np.mean(X_avg_100)
sd = np.std(X_avg_100)
t_100 = np.linspace(mean-3*sd,mean+3*sd,1000)
distribution_100 = norm.pdf(t_100,mean,sd)

plt.hist(X_avg_100, color = 'yellow', density = True)
plt.plot(t_100,distribution_100, color = 'black')
plt.xlabel('Sample Average')
plt.ylabel('Probability (density)')
plt.title('Histogram of sample average at n=100 and corresponding normal distribution')
```
